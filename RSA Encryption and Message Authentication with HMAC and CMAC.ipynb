```python
# 1. Setup given variables and primes
e = 43
p = 13
q = 83
n = p * q

# 2. Calculate the totient (phi) and private key (d)
phi = (p - 1) * (q - 1)
d = pow(e, -1, phi) # d equals 595

# 3. Decrypt the hacker's message
c = [996, 894, 379, 631, 894, 82, 379, 852, 631, 677, 677, 194, 893]
plaintext = ""

for i in c:
    m = pow(i, d, n)     # RSA decryption formula
    plaintext += chr(m)  # Convert to ASCII letters

print(plaintext)
```

```python
import hmac
import hashlib

# Define the secret key and the message
secret_key = b"super_secret_key"
message = b"Cryptography Homework 5 done by Haseeb Ali Syed"

# Generate the HMAC using SHA-256
hmac_result = hmac.new(secret_key, message, hashlib.sha256).hexdigest()

print(f"Message: {message.decode()}")
print(f"HMAC (SHA-256): {hmac_result}")
```

```python
from cryptography.hazmat.primitives import cmac
from cryptography.hazmat.primitives.ciphers import algorithms

# Define the secret key (must be exactly 16, 24, or 32 bytes for AES) and the message
aes_key = b"16byte_secretkey"
message = b"Cryptography Homework 5 done by Haseeb Ali Syed"

# Generate the CMAC using AES
c = cmac.CMAC(algorithms.AES(aes_key))
c.update(message)
cmac_result = c.finalize().hex()

print(f"Message: {message.decode()}")
print(f"CMAC (AES): {cmac_result}")
```
