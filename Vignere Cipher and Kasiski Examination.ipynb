```python
import math

print("--- VIGENERE CIPHER DEMO ---")

def encrypt(text, key):
    result = ""
    idx = 0

    for c in text:
        if c.isalpha():
            shift = ord(key[idx % len(key)].upper()) - 65
            base = 65 if c.isupper() else 97
            result += chr((ord(c) - base + shift) % 26 + base)
            idx += 1
        else:
            result += c

    return result


def decrypt(text, key):
    result = ""
    idx = 0

    for c in text:
        if c.isalpha():
            shift = ord(key[idx % len(key)].upper()) - 65
            base = 65 if c.isupper() else 97
            result += chr((ord(c) - base - shift) % 26 + base)
            idx += 1
        else:
            result += c

    return result


# ---------- SAMPLE INPUT ----------

msg = "CRYPTOGRAPHY IS QUITE IMPORTANT IN CRYPTOCURRENCIES"
key = "MAINS"

print("\n--- Vigenere Encryption & Decryption ---")
print("Plaintext :", msg)
print("Key       :", key)

cipher = encrypt(msg, key)
print("Ciphertext:", cipher)

plain = decrypt(cipher, key)
print("Decrypted :", plain)


# ---------- KASISKI EXAMINATION ----------

print("\n--- KASISKI EXAMINATION ---")


# Remove non-alphabetic characters
clean = ""
for c in cipher:
    if c.isalpha():
        clean += c.upper()

print("\nCiphertext used for Kasiski (cleaned):")
print(clean)

print("\nNote: Using trigrams (3-letter sequences) as required by Kasiski method.")

# 1. Identify repeated sequences
repeats = {}
for i in range(len(clean) - 3):
    seq = clean[i:i+3]
    if seq in repeats:
        repeats[seq].append(i)
    else:
        repeats[seq] = [i]

print("\n--- Step 1: Repeated Sequences ---")
for seq in repeats:
    if len(repeats[seq]) > 1:
        print(f"Sequence '{seq}' found at indices {repeats[seq]}")

# 2. Calculate distances
distances = []
print("\n--- Step 2: Distances Between Repetitions ---")
for seq in repeats:
    if len(repeats[seq]) > 1:
        for i in range(len(repeats[seq]) - 1):
            d = repeats[seq][i+1] - repeats[seq][i]
            distances.append(d)
            print(f"Distance for '{seq}': {d}")

# 3. Common factor frequency (Kasiski heuristic)
factors = {}
for d in distances:
    for i in range(2, d + 1):
        if d % i == 0:
            factors[i] = factors.get(i, 0) + 1

print("\n--- Step 3: Common Factors (Heuristic) ---")
sorted_factors = sorted(factors.items(), key=lambda x: x[1], reverse=True)
for f, count in sorted_factors[:3]:
    print(f"Factor {f} appears {count} times")

# 4. True GCD calculation
gcd_value = distances[0]
for d in distances[1:]:
    gcd_value = math.gcd(gcd_value, d)

print("\n--- Step 4: GCD of Distances ---")
print("Estimated Key Length (GCD):", gcd_value)

# Final interpretation
print("\nConclusion:")
print("The estimated key length allows the Vigenere cipher to be reduced into multiple Caesar ciphers, enabling frequency analysis.")
```
